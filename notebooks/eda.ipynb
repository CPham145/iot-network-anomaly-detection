```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/iot_sample_clean.csv")

# Label distribution
df["Label"].value_counts().plot(kind="bar")
plt.title("Label distribution")
plt.show()

# NaN/Inf sanity
num = df.select_dtypes(include=["number"])
print("NaNs:", int(num.isna().sum().sum()))
print("Infs:", int(np.isinf(num.to_numpy()).sum()))
```
